QUE- **1. What is the difference between interpreted and compiled languages?**

ANS-
An interpreted language executes code line by line using an interpreter, while a compiled language converts the entire source code into machine code before execution. Interpreted languages are easier to debug, whereas compiled languages generally execute faster.

---

QUE- **2. What is exception handling in Python?**

ANS-
Exception handling is a mechanism used to handle runtime errors in a program without stopping its execution. It uses `try`, `except`, `else`, and `finally` blocks.

---

QUE- **3. What is the purpose of the `finally` block in exception handling?**

ANS-
The `finally` block contains code that always executes, whether an exception occurs or not. It is mainly used for cleanup tasks like closing files or releasing resources.

---

QUE- **4. What is logging in Python?**

ANS-
Logging is the process of recording events, errors, warnings, and debugging information during program execution using Python's `logging` module.

---

QUE- **5. What is the significance of the `__del__` method in Python?**

ANS-
The `__del__` method is called when an object is about to be destroyed. It is used to perform cleanup operations such as releasing resources.

---

QUE- **6. What is the difference between `import` and `from ... import` in Python?**

ANS-
`import` imports the entire module, while `from ... import` imports specific functions, classes, or variables from a module.

---

QUE- **7. How can you handle multiple exceptions in Python?**

ANS-
Multiple exceptions can be handled by using multiple `except` blocks or by specifying multiple exception types in a single `except` block.

---

QUE- **8. What is the purpose of the `with` statement when handling files in Python?**

ANS-
The `with` statement automatically manages file resources by opening and closing the file, even if an exception occurs.

---

QUE- **9. What is the difference between multithreading and multiprocessing?**

ANS-
Multithreading uses multiple threads within the same process and shares memory, while multiprocessing uses separate processes with independent memory, making it suitable for CPU-intensive tasks.

---

QUE- **10. What are the advantages of using logging in a program?**

ANS-
Logging helps track program execution, detect errors, simplify debugging, maintain records of events, and monitor application performance.

---

QUE- **11. What is memory management in Python?**

ANS-
Memory management is the process of allocating and releasing memory automatically. Python manages memory using reference counting and garbage collection.

---

QUE- **12. What are the basic steps involved in exception handling in Python?**

ANS-
The basic steps are:
1. Place risky code inside a `try` block.
2. Handle errors using `except`.
3. Use `else` for code that runs if no exception occurs.
4. Use `finally` for cleanup operations.

---

QUE- **13. Why is memory management important in Python?**

ANS-
Memory management prevents memory leaks, improves program performance, ensures efficient resource usage, and enhances application stability.

---

QUE- **14. What is the role of `try` and `except` blocks in exception handling?**

ANS-
The `try` block contains code that may raise an exception, while the `except` block catches and handles the exception to prevent program termination.

---

QUE- **15. How does Python's garbage collection system work?**

ANS-
Python uses reference counting to track object usage and a garbage collector to remove unused objects, especially those involved in circular references.

---

QUE- **16. What is the purpose of the `else` block in exception handling?**

ANS-
The `else` block executes only if no exception occurs in the `try` block. It is used for code that should run after successful execution.

QUE- **17. What are the common logging levels in Python?**

The common logging levels are:
- DEBUG
- INFO
- WARNING
- ERROR
- CRITICAL

QUE- **18. What is the difference between `os.fork()` and multiprocessing in Python?**

ANS-
`os.fork()` creates a child process by duplicating the current process and is mainly available on Unix/Linux systems. The `multiprocessing` module provides a portable and easier way to create and manage multiple processes across different operating systems.

---

QUE- **19. What is the importance of closing a file in Python?**

ANS-
Closing a file releases system resources, saves pending changes, prevents data corruption, and avoids memory leaks.

---

QUE- **20. What is the difference between `file.read()` and `file.readline()` in Python?**

ANS-
`file.read()` reads the entire file or a specified number of characters, while `file.readline()` reads only one line at a time.

QUE- **22. What is the `os` module in Python used for in file handling?**

ANS-
The `os` module provides functions for creating, deleting, renaming, moving, and checking files and directories, as well as interacting with the operating system.

QUE- **23. What are the challenges associated with memory management in Python?**

ANS-
Challenges include memory leaks due to circular references, increased memory usage, performance overhead from garbage collection, and inefficient use of large objects.
